```python
import torch
import torch.nn as nn
import os
from ultralytics import YOLO
import ultralytics.nn.modules.block as block
import ultralytics.nn.tasks as tasks

# 1. ĐỊNH NGHĨA LỚP VIT
class ViT(nn.Module):
    def __init__(self, c1, c2, num_heads=8, mlp_ratio=2.0):
        super().__init__()
        # Đảm bảo c2 chia hết cho num_heads (128 / 8 = 16)
        self.conv = nn.Conv2d(c1, c2, 1, 1) if c1 != c2 else nn.Identity()
        self.norm1 = nn.LayerNorm(c2)
        self.mhsa = nn.MultiheadAttention(embed_dim=c2, num_heads=num_heads, batch_first=True)
        self.norm2 = nn.LayerNorm(c2)
        self.mlp = nn.Sequential(
            nn.Linear(c2, int(c2 * mlp_ratio)), nn.GELU(), nn.Linear(int(c2 * mlp_ratio), c2)
        )

    def forward(self, x):
        x = self.conv(x)
        B, C, H, W = x.shape
        x_flatten = x.flatten(2).transpose(1, 2)
        x_norm = self.norm1(x_flatten)
        attn_out, _ = self.mhsa(x_norm, x_norm, x_norm)
        x_flatten = x_flatten + attn_out
        x_flatten = x_flatten + self.mlp(self.norm2(x_flatten))
        return x_flatten.transpose(1, 2).reshape(B, C, H, W)

# 2. ĐĂNG KÝ MODULE (Khắc phục lỗi KeyError và TypeError)
setattr(block, 'ViT', ViT)
setattr(tasks, 'ViT', ViT)
tasks.ViT = ViT

# ============================
# CẤU HÌNH & RESUME LOGIC
# ============================
device = 'cuda' if torch.cuda.is_available() else 'cpu'
project_name = 'runs_v11s_UAV'
exp_name = 'improved_vit_uav'
# Đường dẫn file last.pt để tiếp tục huấn luyện
checkpoint_path = os.path.join(project_name, exp_name, 'weights', 'last.pt')
yaml_path = '/home/nguyen/Downloads/UAV_Slicing/final_dataset/caiTien_new.yaml'

if os.path.exists(checkpoint_path):
    print(f"🔄 Đang tiếp tục huấn luyện từ checkpoint: {checkpoint_path}")
    model = YOLO(checkpoint_path)
    resume_flag = True
else:
    print("🚀 Bắt đầu huấn luyện mới...")
    model = YOLO(yaml_path)
    model.load('yolo11s.pt')
    resume_flag = False

# ============================
# HUẤN LUYỆN
# ============================
if __name__ == '__main__':
    model.train(
        data='/home/nguyen/Downloads/UAV_Slicing/dataset.yaml',
        epochs=200,
        optimizer='SGD', # Sử dụng SGD như yêu cầu mới của bạn
        momentum=0.9,
        lr0=0.001,
        imgsz=640,
        batch=16,
        device=device,
        project=project_name,
        name=exp_name,
        resume=resume_flag # 🔥 Tự động tiếp tục từ epoch dở dang
    )
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import os
import sys

# Thêm thư mục hiện tại vào path
sys.path.insert(0, os.getcwd())

# ============================
# 1. ĐỊNH NGHĨA LỚP VIT (Cải tiến)
# ============================
class PatchEmbedding(nn.Module):
    """Convert image patches to embeddings."""
    def __init__(self, in_channels, embed_dim, patch_size=4):
        super().__init__()
        self.patch_size = patch_size
        self.proj = nn.Conv2d(in_channels, embed_dim, kernel_size=patch_size, stride=patch_size)
        self.norm = nn.LayerNorm(embed_dim)

    def forward(self, x):
        x = self.proj(x)
        B, C, H, W = x.shape
        x = x.flatten(2).transpose(1, 2)
        x = self.norm(x)
        return x, H, W


class MultiHeadAttention(nn.Module):
    """Multi-Head Self-Attention."""
    def __init__(self, embed_dim, num_heads=4, dropout=0.0):
        super().__init__()
        self.num_heads = num_heads
        self.head_dim = embed_dim // num_heads
        self.scale = self.head_dim ** -0.5
        self.qkv = nn.Linear(embed_dim, embed_dim * 3)
        self.proj = nn.Linear(embed_dim, embed_dim)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        B, N, C = x.shape
        qkv = self.qkv(x).reshape(B, N, 3, self.num_heads, self.head_dim).permute(2, 0, 3, 1, 4)
        q, k, v = qkv[0], qkv[1], qkv[2]
        attn = (q @ k.transpose(-2, -1)) * self.scale
        attn = attn.softmax(dim=-1)
        attn = self.dropout(attn)
        x = (attn @ v).transpose(1, 2).reshape(B, N, C)
        x = self.proj(x)
        return x


class TransformerBlock(nn.Module):
    """Transformer Encoder Block."""
    def __init__(self, embed_dim, num_heads=4, mlp_ratio=4.0, dropout=0.0):
        super().__init__()
        self.norm1 = nn.LayerNorm(embed_dim)
        self.attn = MultiHeadAttention(embed_dim, num_heads, dropout)
        self.norm2 = nn.LayerNorm(embed_dim)
        mlp_hidden = int(embed_dim * mlp_ratio)
        self.mlp = nn.Sequential(
            nn.Linear(embed_dim, mlp_hidden),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(mlp_hidden, embed_dim),
            nn.Dropout(dropout)
        )

    def forward(self, x):
        x = x + self.attn(self.norm1(x))
        x = x + self.mlp(self.norm2(x))
        return x


class ViT(nn.Module):
    """
    Lightweight Vision Transformer for YOLO integration.
    Args:
        in_channels: Input channels (from previous layer)
        out_channels: Output channels
        patch_size: Size of image patches (default: 4)
        embed_dim: Embedding dimension
        depth: Number of transformer blocks (default: 2)
        num_heads: Number of attention heads (default: 4)
    """
    def __init__(self, in_channels, out_channels, patch_size=4, embed_dim=None, depth=2, num_heads=4):
        super().__init__()
        self.in_channels = in_channels
        self.out_channels = out_channels
        embed_dim = embed_dim or out_channels

        self.patch_embed = PatchEmbedding(in_channels, embed_dim, patch_size)
        self.blocks = nn.ModuleList([TransformerBlock(embed_dim, num_heads) for _ in range(depth)])
        self.norm = nn.LayerNorm(embed_dim)
        self.proj_out = nn.Linear(embed_dim, out_channels) if embed_dim != out_channels else nn.Identity()
        self.patch_size = patch_size

    def forward(self, x):
        B, C, H, W = x.shape
        x, pH, pW = self.patch_embed(x)
        for block in self.blocks:
            x = block(x)
        x = self.norm(x)
        x = self.proj_out(x)
        x = x.transpose(1, 2).reshape(B, self.out_channels, pH, pW)
        x = F.interpolate(x, size=(H, W), mode='bilinear', align_corners=False)
        return x


# ============================
# 2. ĐĂNG KÝ MODULE VÀO ULTRALYTICS
# ============================
from ultralytics import YOLO
import ultralytics.nn.modules as modules
import ultralytics.nn.modules.block as block
import ultralytics.nn.tasks as tasks

# Đăng ký ViT vào tất cả các nơi cần thiết
setattr(modules, 'ViT', ViT)
setattr(block, 'ViT', ViT)
setattr(tasks, 'ViT', ViT)

print("✅ ViT module đã được đăng ký thành công!")
```

# Swin Transformer

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import os
import sys

# ============================
# 1. ĐỊNH NGHĨA LỚP SWIN TRANSFORMER
# ============================

class WindowAttention(nn.Module):
    """Window-based Multi-head Self Attention."""
    def __init__(self, dim, window_size, num_heads):
        super().__init__()
        self.dim = dim
        self.window_size = window_size
        self.num_heads = num_heads
        head_dim = dim // num_heads
        self.scale = head_dim ** -0.5

        self.qkv = nn.Linear(dim, dim * 3)
        self.proj = nn.Linear(dim, dim)
        self.softmax = nn.Softmax(dim=-1)

    def forward(self, x):
        B_, N, C = x.shape
        qkv = self.qkv(x).reshape(B_, N, 3, self.num_heads, C // self.num_heads).permute(2, 0, 3, 1, 4)
        q, k, v = qkv[0], qkv[1], qkv[2]

        q = q * self.scale
        attn = (q @ k.transpose(-2, -1))
        attn = self.softmax(attn)

        x = (attn @ v).transpose(1, 2).reshape(B_, N, C)
        x = self.proj(x)
        return x


class SwinTransformerBlock(nn.Module):
    """Swin Transformer Block with shifted window attention."""
    def __init__(self, dim, num_heads, window_size=7, shift_size=0, mlp_ratio=4.0):
        super().__init__()
        self.dim = dim
        self.num_heads = num_heads
        self.window_size = window_size
        self.shift_size = shift_size
        self.mlp_ratio = mlp_ratio

        self.norm1 = nn.LayerNorm(dim)
        self.attn = WindowAttention(dim, window_size, num_heads)
        self.norm2 = nn.LayerNorm(dim)

        mlp_hidden_dim = int(dim * mlp_ratio)
        self.mlp = nn.Sequential(
            nn.Linear(dim, mlp_hidden_dim),
            nn.GELU(),
            nn.Linear(mlp_hidden_dim, dim)
        )

    def forward(self, x):
        H, W = self.H, self.W
        B, L, C = x.shape

        shortcut = x
        x = self.norm1(x)
        x = x.view(B, H, W, C)

        # Pad feature maps to multiples of window size
        pad_l = pad_t = 0
        pad_r = (self.window_size - W % self.window_size) % self.window_size
        pad_b = (self.window_size - H % self.window_size) % self.window_size
        x = F.pad(x, (0, 0, pad_l, pad_r, pad_t, pad_b))
        _, Hp, Wp, _ = x.shape

        # Cyclic shift
        if self.shift_size > 0:
            shifted_x = torch.roll(x, shifts=(-self.shift_size, -self.shift_size), dims=(1, 2))
        else:
            shifted_x = x

        # Partition windows
        x_windows = self.window_partition(shifted_x, self.window_size)
        x_windows = x_windows.view(-1, self.window_size * self.window_size, C)

        # W-MSA/SW-MSA
        attn_windows = self.attn(x_windows)

        # Merge windows
        attn_windows = attn_windows.view(-1, self.window_size, self.window_size, C)
        shifted_x = self.window_reverse(attn_windows, self.window_size, Hp, Wp)

        # Reverse cyclic shift
        if self.shift_size > 0:
            x = torch.roll(shifted_x, shifts=(self.shift_size, self.shift_size), dims=(1, 2))
        else:
            x = shifted_x

        if pad_r > 0 or pad_b > 0:
            x = x[:, :H, :W, :].contiguous()

        x = x.view(B, H * W, C)
        x = shortcut + x
        x = x + self.mlp(self.norm2(x))

        return x

    def window_partition(self, x, window_size):
        B, H, W, C = x.shape
        x = x.view(B, H // window_size, window_size, W // window_size, window_size, C)
        windows = x.permute(0, 1, 3, 2, 4, 5).contiguous().view(-1, window_size, window_size, C)
        return windows

    def window_reverse(self, windows, window_size, H, W):
        B = int(windows.shape[0] / (H * W / window_size / window_size))
        x = windows.view(B, H // window_size, W // window_size, window_size, window_size, -1)
        x = x.permute(0, 1, 3, 2, 4, 5).contiguous().view(B, H, W, -1)
        return x


class SwinTransformer(nn.Module):
    """
    Swin Transformer for YOLO integration.
    Args from YAML: [out_channels]
    """
    def __init__(self, c1, c2, num_heads=4, depth=2, window_size=7):
        super().__init__()
        self.c1 = c1
        self.c2 = c2
        self.window_size = window_size

        # Input projection
        self.input_proj = nn.Conv2d(c1, c2, kernel_size=1)

        # Swin Transformer blocks
        self.blocks = nn.ModuleList()
        for i in range(depth):
            shift_size = 0 if (i % 2 == 0) else window_size // 2
            self.blocks.append(
                SwinTransformerBlock(
                    dim=c2,
                    num_heads=num_heads,
                    window_size=window_size,
                    shift_size=shift_size
                )
            )

        # Output projection
        self.output_proj = nn.Conv2d(c2, c2, kernel_size=1)
        self.norm = nn.LayerNorm(c2)

    def forward(self, x):
        B, C, H, W = x.shape

        # Project input
        x = self.input_proj(x)

        # Flatten to sequence
        x = x.flatten(2).transpose(1, 2)  # (B, H*W, C)

        # Apply Swin blocks
        for block in self.blocks:
            block.H, block.W = H, W
            x = block(x)

        x = self.norm(x)

        # Reshape back to feature map
        x = x.transpose(1, 2).reshape(B, self.c2, H, W)
        x = self.output_proj(x)

        return x
```

```python
# ============================
# 2. ĐĂNG KÝ MODULE VÀO ULTRALYTICS
# ============================
from ultralytics import YOLO
import ultralytics.nn.modules as modules
import ultralytics.nn.modules.block as block
import ultralytics.nn.tasks as tasks
import torch.nn as nn

# Đăng ký SwinTransformer vào các modules
modules.SwinTransformer = SwinTransformer
block.SwinTransformer = SwinTransformer
tasks.SwinTransformer = SwinTransformer

# QUAN TRỌNG: Patch parse_model để xử lý SwinTransformer đúng cách
import ultralytics.nn.tasks as tasks_module
from ultralytics.utils import LOGGER

# Lưu parse_model gốc
_original_parse_model = tasks_module.parse_model

def patched_parse_model(d, ch, verbose=True):
    """
    Patched parse_model to handle SwinTransformer with proper c1 injection.
    Wraps the original parse_model and pre-processes SwinTransformer layers.
    """
    from copy import deepcopy

    # Clone dictionary để không ảnh hưởng original
    d = deepcopy(d)

    # Lấy thông số scale
    scales = d.get("scales", {})
    scale = d.get("scale")
    if scales and not scale:
        scale = list(scales.keys())[0]

    depth_mult = 1.0
    width_mult = 1.0
    max_channels = 1024

    if scales and scale and scale in scales:
        depth_mult, width_mult, max_channels = scales[scale]

    # Tính toán channels cho từng layer
    ch_list = [ch]  # channel 0 = input channels (3)

    def get_ch(f):
        """Get channel count from layer index."""
        if isinstance(f, int):
            return ch_list[f] if f < len(ch_list) else ch_list[-1]
        elif isinstance(f, list):
            return sum(get_ch(x) for x in f)
        return ch_list[-1]

    # Pre-process: Thêm c1 vào args của SwinTransformer
    all_layers = d.get("backbone", []) + d.get("head", [])

    for i, layer in enumerate(all_layers):
        f, n, m, args = layer

        # Tính c1 từ layer trước
        c1 = get_ch(f)

        # Xử lý module type
        if isinstance(m, str) and m == "SwinTransformer":
            # SwinTransformer: args = [c2] -> args = [c1, c2]
            c2 = args[0]
            # Scale c2 theo width_mult
            c2_scaled = int(min(c2, max_channels) * width_mult)
            # Cập nhật args với c1, c2
            layer[3] = [c1, c2_scaled] + args[1:]
            c_out = c2_scaled
        else:
            # Các module khác - ước tính output channels
            if args and isinstance(args[0], int):
                c_out = int(min(args[0], max_channels) * width_mult)
            else:
                c_out = c1

        # Cập nhật channel list
        if isinstance(f, int) and f == -1:
            ch_list.append(c_out)
        elif isinstance(f, list):
            ch_list.append(sum(get_ch(x) for x in f))
        else:
            ch_list.append(c_out)

    # Gọi parse_model gốc với SwinTransformer đã được inject
    return _original_parse_model(d, ch, verbose)

# Thêm SwinTransformer vào global namespace của tasks module
setattr(tasks_module, 'SwinTransformer', SwinTransformer)

# Thay thế parse_model
tasks_module.parse_model = patched_parse_model

print("✅ SwinTransformer module đã được đăng ký thành công!")
```
